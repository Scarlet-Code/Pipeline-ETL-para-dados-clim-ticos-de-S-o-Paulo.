# Análise Exploratória — Dados Climáticos de São Paulo

Este notebook apresenta uma análise exploratória dos dados meteorológicos
coletados pela API OpenWeatherMap e armazenados em formato JSON.

## Objetivos
- Compreender a estrutura dos dados;
- Verificar dimensões e tipos das colunas;
- Identificar valores ausentes;
- Verificar registros duplicados;
- Analisar estatísticas descritivas.

```python

# 1. Importação das bibliotecas

import pandas as pd
import json

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", None)
```

```python

# 2. Carregamento dos dados

path = "../data/raw/weather_data.json"

with open(path, encoding="utf-8") as f:
    data = json.load(f)

df = pd.json_normalize(data)

df.head()
```

```python

# 3. Dimensões, colunas e informações gerais

print("Dimensões:", df.shape)

print("\nColunas:")
print(df.columns.tolist())

df.info()
```

```python

# 4. Verificação de valores nulos

print("Valores nulos por coluna:")
display(df.isnull().sum())
```

```python

# 5. Verificação de registros duplicados

print(
    "Registros duplicados:",
    df.duplicated(subset=["id", "dt"]).sum()
)
```

```python

# 6. Estatísticas descritivas

display(df.describe().T)
```
